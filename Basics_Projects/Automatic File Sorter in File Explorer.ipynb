# Automatic File Sorter in File Explorer

```python
import os, shutil
```

```python
path = r"D:/OMKAR/files/notes/"
```

```python
file_name = os.listdir(path)
```

```python
folder_names = ['csv files', 'images', 'text files', 'excel files', 'video files', 'python files']

for loop in range (0,6):
    if not os.path.exists(path + folder_names[loop]):
        print(path + folder_names[loop])
        os.makedirs(path + folder_names[loop])
```

```python
for file in file_name :
    if ".csv" in file and not os.path.exists(path + "csv files/" + file):
        shutil.move(path + file, path + "csv files/" + file)
    elif any(ext in file for ext in ('.png', '.jpg', '.jpeg', '.gif')) and not os.path.exists(path + "images/" + file):
        shutil.move(path + file, path + "images/" + file)
    elif any(ext in file for ext in ('.txt', '.doc', '.docx', '.pdf')) and not os.path.exists(path + "text files/" + file):
        shutil.move(path + file, path + "text files/" + file)
    elif any(ext in file for ext in ('.xlsx', '.xls')) and not os.path.exists(path + "excel files/" + file):
        shutil.move(path + file, path + "excel files/" + file)
    elif any(ext in file for ext in ('.mp4', '.mkv', '.avi')) and not os.path.exists(path + "video files/" + file):
        shutil.move(path + file, path + "video files/" + file)
    elif any(ext in file for ext in ('.py', '.ipynb')) and not os.path.exists(path + "python files/" + file):
        shutil.move(path + file, path + "python files/" + file)
    else :
        print("There is file that not move")
```
