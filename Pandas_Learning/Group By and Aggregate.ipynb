# Group By and Aggregate

```python
import pandas as pd 
```

```python
df = pd.read_csv(r"D:\OMKAR\Python\Pandas_Learning\Raw Data\Flavors.csv")
df
```

```python
group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
df.groupby('Base Flavor').mean(numeric_only=True)
```

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
```

```python
df.groupby('Base Flavor').sum(numeric_only = True)
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean', 'count', 'max', 'min', 'sum']})
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean', 'count', 'max', 'min', 'sum'], 'Texture Rating': ['mean', 'count', 'max', 'min', 'sum']})
```

```python
df.groupby(['Base Flavor', 'Liked']).sum(numeric_only = True)
```

```python
df.groupby(['Base Flavor', 'Liked']).agg({'Flavor Rating': ['mean', 'count', 'max', 'min', 'sum']})
```

```python
df.groupby('Base Flavor').describe()
```
